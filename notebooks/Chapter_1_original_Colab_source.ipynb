```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

days = 200
price = 100 + np.cumsum(np.random.normal(0, 1, days))

df = pd.DataFrame({
    "Day": np.arange(days),
    "Actual_Price": price
})

df.head()
```

```python
# Simulated predictions with small numerical error
predicted_price = df["Actual_Price"] + np.random.normal(0, 0.5, days)

df["Predicted_Price"] = predicted_price

# Calculate MSE
mse = np.mean((df["Predicted_Price"] - df["Actual_Price"])**2)
print("MSE:", mse)

# Compute market direction
df["Actual_Change"] = df["Actual_Price"].diff()
df["Pred_Change"] = df["Predicted_Price"].diff()

# Trading signal
df["Position"] = np.where(df["Pred_Change"] > 0, 1, -1)

# Profit
df["Profit"] = df["Position"] * df["Actual_Change"]

print("Total Trading Profit:", df["Profit"].sum())
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["Actual_Price"], label="Actual Price")
plt.plot(df["Predicted_Price"], label="Predicted Price")
plt.title("Low MSE but Poor Trading Profit")
plt.legend()
plt.show()
```

```python
actual = 100
predictions = [105, 95]

errors = [(p - actual)**2 for p in predictions]

print("Prediction:", predictions[0], "MSE:", errors[0])
print("Prediction:", predictions[1], "MSE:", errors[1])
```

```python
labels = ["Overprediction (105)", "Underprediction (95)"]
values = errors

plt.bar(labels, values)
plt.title("Symmetric MSE Penalty")
plt.ylabel("MSE Loss")
plt.show()
```

```python
transaction_cost = 0.5

df["Profit_No_Cost"] = df["Profit"]
df["Profit_With_Cost"] = df["Profit"] - transaction_cost

print("Profit without cost:", df["Profit_No_Cost"].sum())
print("Profit with cost:", df["Profit_With_Cost"].sum())
```

```python
profits = [
    df["Profit_No_Cost"].sum(),
    df["Profit_With_Cost"].sum()
]

labels = ["No Transaction Cost", "With Transaction Cost"]

plt.bar(labels, profits)
plt.title("Trading Friction Impact")
plt.ylabel("Total Profit")
plt.show()
```
